```python
df['customer_rating'] = df['customer_rating'].fillna(
    df['customer_rating'].mean()
)
```

```python
df['payment_method'] = df['payment_method'].fillna(
    'Unknown'
)
```

```python
df['delivery_status'] = df['delivery_status'].str.capitalize()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.to_csv(
    "clean_food_delivery.csv",
    index=False
)
```
